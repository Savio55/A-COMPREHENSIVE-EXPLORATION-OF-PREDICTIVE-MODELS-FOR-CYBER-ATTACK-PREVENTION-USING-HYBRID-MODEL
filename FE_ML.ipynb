# Create New Port Based Features

PACKET SIZE DISTRIBUTION FEATURES

```python
# Byte volume asymmetry
df['byte_volume_ratio'] = df['Total_Length_of_Fwd_Packets'] / (df['Total_Length_of_Bwd_Packets'] + 1)
df['byte_volume_diff'] = df['Total_Length_of_Fwd_Packets'] - df['Total_Length_of_Bwd_Packets']

# Direction dominance indicators
df['forward_dominant'] = (df['packet_count_ratio'] > 2).astype(int)
df['backward_dominant'] = (df['packet_count_ratio'] < 0.5).astype(int)
df['balanced_traffic'] = ((df['packet_count_ratio'] >= 0.5) & (df['packet_count_ratio'] <= 2)).astype(int)
```

TIMING AND RATE FEATURES

```python
# Flow rate calculations
df['bytes_per_second'] = (df['Total_Length_of_Fwd_Packets'] + df['Total_Length_of_Bwd_Packets']) / (df['Flow_Duration'] + 1)
df['packets_per_second'] = (df['Total_Fwd_Packets'] + df['Total_Backward_Packets']) / (df['Flow_Duration'] + 1)

# Inter-arrival time features
df['iat_forward_backward_ratio'] = df['Fwd_IAT_Mean'] / (df['Bwd_IAT_Mean'] + 1)
df['total_iat_variation'] = df['Flow_IAT_Std'] / (df['Flow_IAT_Mean'] + 1)
df['fwd_iat_variation'] = df['Fwd_IAT_Std'] / (df['Fwd_IAT_Mean'] + 1)
df['bwd_iat_variation'] = df['Bwd_IAT_Std'] / (df['Bwd_IAT_Mean'] + 1)

# Burstiness indicators
df['is_bursty'] = (df['Flow_IAT_Std'] > df['Flow_IAT_Mean']).astype(int)
df['burstiness_score'] = (df['Flow_IAT_Std'] - df['Flow_IAT_Mean']) / (df['Flow_IAT_Std'] + df['Flow_IAT_Mean'] + 1)

# Timing consistency
df['iat_consistency'] = (df['Flow_IAT_Max'] - df['Flow_IAT_Min']) / (df['Flow_IAT_Mean'] + 1)
df['fwd_bwd_iat_similarity'] = abs(df['Fwd_IAT_Mean'] - df['Bwd_IAT_Mean']) / ((df['Fwd_IAT_Mean'] + df['Bwd_IAT_Mean']) / 2 + 1)
```

PROTOCOL FLAG PATTERN FEATURES

```python
# TCP flag combinations
df['syn_only'] = ((df['SYN_Flag_Count'] > 0) &
                  (df['ACK_Flag_Count'] == 0) &
                  (df['FIN_Flag_Count'] == 0)).astype(int)

df['syn_ack'] = ((df['SYN_Flag_Count'] > 0) &
                 (df['ACK_Flag_Count'] > 0)).astype(int)

df['fin_only'] = ((df['FIN_Flag_Count'] > 0) &
                  (df['SYN_Flag_Count'] == 0) &
                  (df['ACK_Flag_Count'] == 0)).astype(int)

df['rst_present'] = (df['RST_Flag_Count'] > 0).astype(int)

# Flag density and ratios
df['flag_density'] = (df['FIN_Flag_Count'] + df['SYN_Flag_Count'] +
                      df['RST_Flag_Count'] + df['PSH_Flag_Count'] +
                      df['ACK_Flag_Count'] + df['URG_Flag_Count']) / (df['Total_Fwd_Packets'] + df['Total_Backward_Packets'] + 1)

df['syn_ack_ratio'] = df['SYN_Flag_Count'] / (df['ACK_Flag_Count'] + 1)
df['fin_ack_ratio'] = df['FIN_Flag_Count'] / (df['ACK_Flag_Count'] + 1)

# Unusual flag combinations
df['urgent_flag_present'] = ((df['Fwd_URG_Flags'] > 0) | (df['Bwd_URG_Flags'] > 0)).astype(int)
df['push_flag_present'] = ((df['Fwd_PSH_Flags'] > 0) | (df['Bwd_PSH_Flags'] > 0)).astype(int)
df['urgent_push_combo'] = ((df['urgent_flag_present'] == 1) & (df['push_flag_present'] == 1)).astype(int)
```

CONNECTION BEHAVIOR FEATURES

```python
# Connection lifespan categories
df['is_short_flow'] = (df['Flow_Duration'] < 1000).astype(int)
df['is_medium_flow'] = ((df['Flow_Duration'] >= 1000) & (df['Flow_Duration'] < 60000)).astype(int)
df['is_long_flow'] = (df['Flow_Duration'] >= 60000).astype(int)

# Packet count categories
df['low_packet_count'] = ((df['Total_Fwd_Packets'] + df['Total_Backward_Packets']) < 5).astype(int)
df['medium_packet_count'] = ((df['Total_Fwd_Packets'] + df['Total_Backward_Packets'] >= 5) &
                            (df['Total_Fwd_Packets'] + df['Total_Backward_Packets'] <= 100)).astype(int)
df['high_packet_count'] = ((df['Total_Fwd_Packets'] + df['Total_Backward_Packets']) > 100).astype(int)

# Session completeness indicators
df['incomplete_session'] = ((df['SYN_Flag_Count'] > 0) &
                            (df['FIN_Flag_Count'] == 0) &
                            (df['RST_Flag_Count'] == 0)).astype(int)

df['abnormal_termination'] = ((df['RST_Flag_Count'] > 0) &
                              (df['FIN_Flag_Count'] == 0)).astype(int)

df['normal_termination'] = ((df['FIN_Flag_Count'] > 0) &
                            (df['RST_Flag_Count'] == 0)).astype(int)
```

WINDOW SIZE AND HEADER FEATURES

```python
# Window size anomalies
df['zero_window_forward'] = (df['Init_Win_bytes_forward'] == 0).astype(int)
df['zero_window_backward'] = (df['Init_Win_bytes_backward'] == 0).astype(int)
df['zero_window_any'] = ((df['Init_Win_bytes_forward'] == 0) |
                         (df['Init_Win_bytes_backward'] == 0)).astype(int)

df['large_window_forward'] = (df['Init_Win_bytes_forward'] > 65535).astype(int)
df['large_window_backward'] = (df['Init_Win_bytes_backward'] > 65535).astype(int)
df['large_window_any'] = ((df['Init_Win_bytes_forward'] > 65535) |
                          (df['Init_Win_bytes_backward'] > 65535)).astype(int)

# Negative window size (anomalous)
df['negative_window'] = ((df['Init_Win_bytes_forward'] < 0) |
                         (df['Init_Win_bytes_backward'] < 0)).astype(int)

# Header length features
df['header_length_ratio'] = df['Fwd_Header_Length'] / (df['Bwd_Header_Length'] + 1)
df['total_header_length'] = df['Fwd_Header_Length'] + df['Bwd_Header_Length']
df['header_length_diff'] = df['Fwd_Header_Length'] - df['Bwd_Header_Length']

# Duplicate header length check
df['header_length_match'] = (df['Fwd_Header_Length'] == df['Fwd_Header_Length_1']).astype(int)
df['header_length_discrepancy'] = abs(df['Fwd_Header_Length'] - df['Fwd_Header_Length_1'])
```

BULK TRANSFER AND SEGMENT FEATURES

```python
# Bulk transfer indicators
df['has_bulk_transfer'] = ((df['Fwd_Avg_Bytes_Bulk'] > 0) |
                           (df['Bwd_Avg_Bytes_Bulk'] > 0)).astype(int)

df['bulk_transfer_size'] = df['Fwd_Avg_Bytes_Bulk'] + df['Bwd_Avg_Bytes_Bulk']
df['bulk_transfer_rate'] = df['Fwd_Avg_Bulk_Rate'] + df['Bwd_Avg_Bulk_Rate']

# Bulk asymmetry
df['bulk_direction_ratio'] = df['Fwd_Avg_Bytes_Bulk'] / (df['Bwd_Avg_Bytes_Bulk'] + 1)
df['bulk_packet_ratio'] = df['Fwd_Avg_Packets_Bulk'] / (df['Bwd_Avg_Packets_Bulk'] + 1)

# Segment size features
df['segment_size_ratio'] = df['Avg_Fwd_Segment_Size'] / (df['Avg_Bwd_Segment_Size'] + 1)
df['segment_size_diff'] = df['Avg_Fwd_Segment_Size'] - df['Avg_Bwd_Segment_Size']
df['total_segment_size'] = df['Avg_Fwd_Segment_Size'] + df['Avg_Bwd_Segment_Size']

# Average packet size validation
df['calculated_avg_packet_size'] = (df['Total_Length_of_Fwd_Packets'] + df['Total_Length_of_Bwd_Packets']) / (
    df['Total_Fwd_Packets'] + df['Total_Backward_Packets'] + 1)
df['avg_packet_size_diff'] = abs(df['Average_Packet_Size'] - df['calculated_avg_packet_size'])
```

SUBFLOW AND DATA TRANSFER FEATURES

```python
# Subflow ratios
df['subflow_packet_ratio'] = df['Subflow_Fwd_Packets'] / (df['Subflow_Bwd_Packets'] + 1)
df['subflow_byte_ratio'] = df['Subflow_Fwd_Bytes'] / (df['Subflow_Bwd_Bytes'] + 1)

# Subflow to total ratios
df['subflow_fwd_packet_ratio'] = df['Subflow_Fwd_Packets'] / (df['Total_Fwd_Packets'] + 1)
df['subflow_bwd_packet_ratio'] = df['Subflow_Bwd_Packets'] / (df['Total_Backward_Packets'] + 1)

# Data packet features
df['data_packet_ratio'] = df['act_data_pkt_fwd'] / (df['Total_Fwd_Packets'] + 1)
df['has_data_packets'] = (df['act_data_pkt_fwd'] > 0).astype(int)

# Segment size features
df['min_segment_size_anomaly'] = (df['min_seg_size_forward'] < 20).astype(int)
df['segment_size_consistency'] = df['min_seg_size_forward'] / (df['Avg_Fwd_Segment_Size'] + 1)
```

ACTIVITY AND IDLE PATTERN FEATURES

```python
# Activity patterns
df['activity_ratio'] = df['Active_Mean'] / (df['Idle_Mean'] + 1)
df['activity_variation'] = df['Active_Std'] / (df['Active_Mean'] + 1)
df['idle_variation'] = df['Idle_Std'] / (df['Idle_Mean'] + 1)

# Peak activity indicators
df['high_activity'] = (df['Active_Max'] > df['Active_Mean'] * 3).astype(int)
df['low_activity'] = (df['Active_Max'] < df['Active_Mean'] * 0.5).astype(int)

df['long_idle'] = (df['Idle_Max'] > 1000000).astype(int)
df['short_idle'] = (df['Idle_Max'] < 1000).astype(int)

# Activity consistency
df['activity_range'] = df['Active_Max'] - df['Active_Min']
df['idle_range'] = df['Idle_Max'] - df['Idle_Min']
df['activity_idle_ratio'] = df['activity_range'] / (df['idle_range'] + 1)

# Active/Idle correlation
df['active_idle_correlation'] = (df['Active_Mean'] > df['Idle_Mean']).astype(int)
```

 RATE AND THROUGHPUT FEATURES

```python
# Forward vs backward rates
df['fwd_bwd_packet_rate_ratio'] = df['Fwd_Packets_s'] / (df['Bwd_Packets_s'] + 1)
df['fwd_bwd_byte_rate_ratio'] = (df['Total_Length_of_Fwd_Packets'] / df['Flow_Duration'] + 1) / (
    df['Total_Length_of_Bwd_Packets'] / df['Flow_Duration'] + 1 + 1)

# Rate anomalies
df['high_flow_rate'] = (df['Flow_Packets_s'] > df['Flow_Packets_s'].quantile(0.95)).astype(int)
df['low_flow_rate'] = (df['Flow_Packets_s'] < df['Flow_Packets_s'].quantile(0.05)).astype(int)

df['high_byte_rate'] = (df['Flow_Bytes_s'] > df['Flow_Bytes_s'].quantile(0.95)).astype(int)
df['low_byte_rate'] = (df['Flow_Bytes_s'] < df['Flow_Bytes_s'].quantile(0.05)).astype(int)

# Rate consistency
df['packet_rate_consistency'] = df['Fwd_Packets_s'] / (df['Flow_Packets_s'] + 1)
df['byte_rate_consistency'] = (df['Total_Length_of_Fwd_Packets'] / df['Flow_Duration']) / (df['Flow_Bytes_s'] + 1)
```

INTERACTION AND RATIO FEATURES

```python
# Packet size * rate interactions
df['size_rate_interaction'] = df['Avg_Fwd_Segment_Size'] * df['Fwd_Packets_s']
df['duration_packet_interaction'] = df['Flow_Duration'] * df['Total_Fwd_Packets']
df['duration_byte_interaction'] = df['Flow_Duration'] * (df['Total_Length_of_Fwd_Packets'] + df['Total_Length_of_Bwd_Packets'])

# Flag * packet count interactions
df['syn_packet_ratio'] = df['SYN_Flag_Count'] / (df['Total_Fwd_Packets'] + 1)
df['ack_packet_ratio'] = df['ACK_Flag_Count'] / (df['Total_Fwd_Packets'] + 1)
df['fin_packet_ratio'] = df['FIN_Flag_Count'] / (df['Total_Fwd_Packets'] + 1)

# Timing * size interactions
df['iat_size_ratio'] = df['Fwd_IAT_Mean'] / (df['Fwd_Packet_Length_Mean'] + 1)
df['duration_size_ratio'] = df['Flow_Duration'] / (df['Total_Length_of_Fwd_Packets'] + df['Total_Length_of_Bwd_Packets'] + 1)

# Subflow to main flow ratios
df['subflow_main_packet_ratio'] = (df['Subflow_Fwd_Packets'] + df['Subflow_Bwd_Packets']) / (
    df['Total_Fwd_Packets'] + df['Total_Backward_Packets'] + 1)
df['subflow_main_byte_ratio'] = (df['Subflow_Fwd_Bytes'] + df['Subflow_Bwd_Bytes']) / (
    df['Total_Length_of_Fwd_Packets'] + df['Total_Length_of_Bwd_Packets'] + 1)
```

 NORMALIZATION AND TRANSFORMATION FEATURES

```python
import numpy as np

# Log transformations for highly skewed features
df['log_flow_duration'] = np.log1p(df['Flow_Duration'])
df['log_total_fwd_packets'] = np.log1p(df['Total_Fwd_Packets'])
df['log_total_bwd_packets'] = np.log1p(df['Total_Backward_Packets'])
df['log_total_fwd_bytes'] = np.log1p(df['Total_Length_of_Fwd_Packets'])
df['log_total_bwd_bytes'] = np.log1p(df['Total_Length_of_Bwd_Packets'])
df['log_flow_bytes_s'] = np.log1p(df['Flow_Bytes_s'])
df['log_flow_packets_s'] = np.log1p(df['Flow_Packets_s'])

# Square root transformations
df['sqrt_flow_duration'] = np.sqrt(df['Flow_Duration'] + 1)
df['sqrt_total_packets'] = np.sqrt(df['Total_Fwd_Packets'] + df['Total_Backward_Packets'] + 1)

# Reciprocal transformations (for very large values)
df['reciprocal_flow_duration'] = 1 / (df['Flow_Duration'] + 1)
df['reciprocal_packet_count'] = 1 / (df['Total_Fwd_Packets'] + df['Total_Backward_Packets'] + 1)

# Binning for categorical transformation
df['flow_duration_bin'] = pd.cut(df['Flow_Duration'],
                                  bins=[0, 100, 1000, 10000, 100000, float('inf')],
                                  labels=['micro', 'short', 'medium', 'long', 'very_long'])

df['packet_count_bin'] = pd.cut(df['Total_Fwd_Packets'] + df['Total_Backward_Packets'],
                                 bins=[0, 5, 20, 100, 500, float('inf')],
                                 labels=['very_low', 'low', 'medium', 'high', 'very_high'])

# Z-score normalization for key features
for col in ['Flow_Duration', 'Total_Fwd_Packets', 'Total_Length_of_Fwd_Packets',
            'Flow_Bytes_s', 'Flow_Packets_s']:
    df[f'zscore_{col}'] = (df[col] - df[col].mean()) / (df[col].std() + 1e-10)
```
